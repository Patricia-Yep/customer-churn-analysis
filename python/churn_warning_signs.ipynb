```python
import pandas as pd

df = pd.read_csv("../data/customer_churn_business_dataset_cleaned.csv")
df.shape
```

```python
df["churn"].mean()
```

```python
df["tenure_group"]=pd.cut(
    df["tenure_months"],
    bins=[0,6,12,24,48,60],
    labels=["0-6 months", "7-12 months", "1-2 years", "2-4 years", "4-5 years"]
)

df[["tenure_months", "tenure_group"]].head(10)
```

```python
df.groupby("tenure_group", observed=True)["churn"].agg(["mean", "count"])
```

```python
df.groupby("csat_score")["churn"].agg(["mean", "count"])
```

```python
df.groupby("payment_failures")["churn"].agg(["mean", "count"])
```

```python
df["failures_2plus"] = df["payment_failures"] >= 2

df.groupby("failures_2plus")["churn"].agg(["mean", "count"])
```

```python
df["login_group"]=pd.qcut(df["monthly_logins"],4)

df.groupby("login_group", observed=True)["churn"].agg(["mean","count"])
```

```python
df.groupby(pd.cut(df["monthly_logins"], 4), observed=True)["churn"].agg(["mean", "count"])
```

```python
churned = df[df["churn"] ==1]
churned["monthly_fee"].sum()
```

```python
churned["monthly_fee"].sum()/df["monthly_fee"].sum()
```

```python
df.groupby(
    pd.cut(df["last_login_days_ago"], bins=[-1, 7, 14, 30, 80]),
    observed=True
)["churn"].agg(["mean", "count"])
```
